```python
import pandas as pd

df = pd.read_csv("../sentiment analysis.csv")

df.head()
```

```python
df.shape

df.columns

df.info()

df.isnull().sum()
```

```python
df.dropna(inplace=True)

df.drop_duplicates(inplace=True)

df.shape
```

```python
import re

def clean_text(text):
    
    text = str(text).lower()
    
    text = re.sub(r'http\S+', '', text)
    
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    
    text = re.sub(r'\s+', ' ', text)
    
    return text.strip()
```

```python
df['clean_text'] = df['Sub Category'].apply(clean_text)

df[['Sub Category','clean_text']].head()
```

```python
import nltk

nltk.download('punkt')

nltk.download('stopwords')
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    
    words = word_tokenize(text)
    
    filtered = [word for word in words if word not in stop_words]
    
    return " ".join(filtered)

df['clean_text'] = df['clean_text'].apply(remove_stopwords)

df[['Sub Category','clean_text']].head()
```

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    
    words = text.split()
    
    words = [lemmatizer.lemmatize(word) for word in words]
    
    return " ".join(words)

df['clean_text'] = df['clean_text'].apply(lemmatize_text)

df[['Sub Category','clean_text']].head()
```

Topic Modeling & Department Prediction
- STEP 1 : Features and Target 

```python
X = df['clean_text']

y = df['Category']
```

- STEP 2 : TF-IDF Vectorization 

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=5000)

X_vectorized = tfidf.fit_transform(X)

print(X_vectorized.shape)
```

- STEP 3- Train Test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_vectorized,
    y,
    test_size=0.2,
    random_state=42
)

print(X_train.shape)

print(X_test.shape)
```

- Logistic Regression 

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression()

lr_model.fit(X_train, y_train)

lr_pred = lr_model.predict(X_test)
from sklearn.metrics import accuracy_score

lr_accuracy = accuracy_score(y_test, lr_pred)

print("Logistic Regression Accuracy :", lr_accuracy)
from sklearn.metrics import classification_report
  
print(classification_report(y_test, lr_pred))
```

```python
from sklearn.metrics import confusion_matrix

lr_cm = confusion_matrix(y_test, lr_pred)

print("Logistic Regression Confusion Matrix:\n")

print(lr_cm)
```

- Random Forest 


```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
```

```python
rf_accuracy = accuracy_score(y_test, rf_pred)

print("Random Forest Accuracy :", rf_accuracy)
```

```python
print(classification_report(y_test, rf_pred))
```

```python
rf_cm = confusion_matrix(y_test, rf_pred)

print("Random Forest Confusion Matrix:\n")

print(rf_cm)
```

- cross validation of logistic and random forest 

```python
from sklearn.model_selection import cross_val_score


lr_scores = cross_val_score(
    lr_model,
    X_vectorized,
    y,
    cv=5
)

print("LR Cross Validation Scores :", lr_scores)

print("Average LR Accuracy :", lr_scores.mean())
```

```python
rf_scores = cross_val_score(
    rf_model,
    X_vectorized,
    y,
    cv=5
)

print("RF Cross Validation Scores :", rf_scores)

print("Average RF Accuracy :", rf_scores.mean())
```

```python
sample = [
    "street light not working"
]

sample_vector = tfidf.transform(sample)

# Logistic Regression Prediction

lr_result = lr_model.predict(sample_vector)

print("LR Prediction :", lr_result[0])

# Random Forest Prediction

rf_result = rf_model.predict(sample_vector)

print("RF Prediction :", rf_result[0])
```
